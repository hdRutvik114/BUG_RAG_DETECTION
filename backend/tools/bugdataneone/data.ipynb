```python
import json
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")


def load_dataset(file_path):
    """Loads dataset from JSON, Parquet, or CSV."""
    if file_path.endswith(".json") or file_path.endswith(".jsonl"):
        return pd.read_json(file_path)
    elif file_path.endswith(".parquet"):
        return pd.read_parquet(file_path)
    elif file_path.endswith(".csv"):
        return pd.read_csv(file_path)
    else:
        raise ValueError("Unsupported file format")


def analyze_dataset_quality(df):
    print("==================================================")
    print("          DATASET DIAGNOSTIC REPORT              ")
    print("==================================================")

    # 1. Calculate Line Lengths
    df["buggy_loc"] = df["buggy_code"].apply(
        lambda x: len(str(x).splitlines()) if pd.notna(x) else 0
    )
    df["fixed_loc"] = df["fixed_code"].apply(
        lambda x: len(str(x).splitlines()) if pd.notna(x) else 0
    )
    df["loc_ratio"] = df["fixed_loc"] / (df["buggy_loc"] + 1e-5)

    print(f"\n[+] Total Samples: {len(df)}")
    print(f"[+] Median Buggy LOC: {df['buggy_loc'].median():.1f}")
    print(f"[+] Median Fixed LOC: {df['fixed_loc'].median():.1f}")
    print(f"[+] Median Fixed/Buggy Size Ratio: {df['loc_ratio'].median():.1f}x")

    # Mismatch Flag: Fixed code is > 5x larger than Buggy hunk
    severe_mismatches = df[df["loc_ratio"] > 5]
    mismatch_pct = (len(severe_mismatches) / len(df)) * 100
    print(
        f"🚨 Severe Size Mismatches (Fixed > 5x Buggy): {len(severe_mismatches)} ({mismatch_pct:.1f}%)"
    )

    # ----------------------------------------------------
    # PLOTTING
    # ----------------------------------------------------
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))

    # Plot 1: Line Count Mismatch Distribution
    sns.boxplot(
        data=df[["buggy_loc", "fixed_loc"]], ax=axes[0, 0], palette="Set2"
    )
    axes[0, 0].set_yscale("log")
    axes[0, 0].set_title("1. Buggy vs Fixed Code Length (Log Scale)")
    axes[0, 0].set_ylabel("Lines of Code (LOC)")
    axes[0, 0].set_xticklabels(["Buggy (Hunk)", "Fixed (Full File)"])

    # Plot 2: Ratio Distribution (Disparity)
    sns.histplot(
        df["loc_ratio"], bins=30, kde=True, ax=axes[0, 1], color="crimson"
    )
    axes[0, 1].set_xlim(0, 50)
    axes[0, 1].set_title("2. Size Disparity Ratio (Fixed_LOC / Buggy_LOC)")
    axes[0, 1].set_xlabel("Ratio (Higher = Worse Mismatch)")

    # Plot 3: Bug Type/Category Distribution (if present)
    bug_col = next(
        (col for col in ["bug_type", "category", "label"] if col in df.columns),
        None,
    )
    if bug_col:
        sns.countplot(
            data=df,
            y=bug_col,
            ax=axes[1, 0],
            order=df[bug_col].value_counts().index,
            palette="viridis",
        )
        axes[1, 0].set_title(f"3. Label Distribution ({bug_col})")
        axes[1, 0].set_xlabel("Count")
    else:
        axes[1, 0].text(
            0.5, 0.5, "No label column found", ha="center", va="center"
        )
        axes[1, 0].set_title("3. Label Distribution (Missing)")

    # Plot 4: Scatter Plot of Buggy vs Fixed
    sns.scatterplot(
        data=df,
        x="buggy_loc",
        y="fixed_loc",
        alpha=0.5,
        ax=axes[1, 1],
        color="teal",
    )
    axes[1, 1].set_title("4. Buggy vs. Fixed Size Correlation")
    axes[1, 1].set_xlabel("Buggy LOC")
    axes[1, 1].set_ylabel("Fixed LOC")

    plt.tight_layout()
    plt.show()

    # ----------------------------------------------------
    # SAMPLE INSPECTION
    # ----------------------------------------------------
    print("\n==================================================")
    print("          EXAMPLE OF SEVERE MISMATCH              ")
    print("==================================================")
    if len(severe_mismatches) > 0:
        sample = severe_mismatches.iloc[0]
        print(f"Repository: {sample.get('repo', 'N/A')}")
        print(f"Commit Hash: {sample.get('commit_hash', 'N/A')}")
        print(
            f"Buggy Code LOC: {sample['buggy_loc']} | Fixed Code LOC: {sample['fixed_loc']}"
        )
        print("\n--- BUGGY CODE SNIPPET ---")
        print("\n".join(str(sample["buggy_code"]).splitlines()[:10]))
        print("... [truncated]")
        print("\n--- FIXED CODE SNIPPET ---")
        print("\n".join(str(sample["fixed_code"]).splitlines()[:10]))
        print("... [truncated]")


# Usage: Replace 'dataset.parquet' or 'dataset.json' with your file
df = load_dataset(r'C:\Users\YATIRAJ\OneDrive\Desktop\bugdataneone\dataset_enriched.csv')
analyze_dataset_quality(df)
```
